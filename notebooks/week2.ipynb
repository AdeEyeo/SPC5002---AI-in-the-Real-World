```python
import json
import pandas as pd

web = pd.read_csv("../data/sources/web_orders_2025.csv")
calls = pd.json_normalize(json.load(open("../data/sources/phone_orders_2025.json", encoding="utf-8"))["calls"])
crm = pd.read_excel("../data/sources/crm_customers.xlsx", skiprows=2)
week1_cols = pd.read_csv("../data/retail_orders_week1.csv", nrows=0).columns

for name, df in [("web", web), ("phone", calls), ("crm", crm)]:
    print(name, df.shape)
    print(df.dtypes, "\n")
print("Week 1 columns:", list(week1_cols))
```
